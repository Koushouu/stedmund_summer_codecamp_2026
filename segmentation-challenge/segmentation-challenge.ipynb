```python
from skimage.io import imread
import matplotlib.pyplot as plt

filepath = r'..\example_images\example_cells.tif'
img = imread(filepath)

print("Loaded array is of type:", type(img))
print("Loaded array has shape:", img.shape)
print("Loaded values are of type:", img.dtype)

plt.imshow(img, interpolation='none', cmap='gray')
```

```python
import numpy as np
from sklearn.metrics import adjusted_rand_score

ground_truth_mask = imread(r'gt.tif')
test_mask = imread(r'test.tif')
random_mask = np.random.randint(0, 300, size=ground_truth_mask.shape)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(random_mask, cmap='prism', alpha=0.8)
axes[0].set_title('Random Mask')
axes[1].imshow(test_mask, cmap='prism', alpha=0.8)
axes[1].set_title('Test Mask')
axes[2].imshow(ground_truth_mask, cmap='prism', alpha=0.8)
axes[2].set_title('Ground Truth Mask')
score = adjusted_rand_score(ground_truth_mask.flatten(), random_mask.flatten())
print(f"Adjusted Rand Index score: {score:.4f}")
score = adjusted_rand_score(ground_truth_mask.flatten(), test_mask.flatten())
print(f"Adjusted Rand Index score: {score:.4f}")
```
